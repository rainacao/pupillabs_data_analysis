```python
from pathlib import Path
import msgpack
import numpy as np
import pandas as pd

def load_pldata(rec_dir, topic):
    rec_dir = Path(rec_dir)
    rows = []
    with open(rec_dir / f"{topic}.pldata", "rb") as f:
        for _topic, payload in msgpack.Unpacker(f, raw=False, use_list=False):
            rows.append(msgpack.unpackb(payload, raw=False, use_list=False))
    return rows

gaze = load_pldata("recordings/004", "gaze")
gaze_df = pd.DataFrame(
    {
        "timestamp": [g["timestamp"] for g in gaze],
        "x": [g["norm_pos"][0] for g in gaze],
        "y": [g["norm_pos"][1] for g in gaze],
        "confidence": [g["confidence"] for g in gaze],
    }
)
world_ts = np.load("recordings/004/world_timestamps.npy")  # one timestamp per world video frame
```
